```python
# File uploader
import base64
from ipywidgets import FileUpload
from IPython.display import display

uploader = FileUpload(accept='.pdf', multiple=False)
display(uploader)
```

```python
import io
from pypdf import PdfReader
from langchain_typesafe import Noul, TypeSafeClassifier

def get_uploaded_pdf(uploader):
    """Return (filename, bytes) from a FileUpload widget (ipywidgets 7 or 8)."""
    val = uploader.value
    if not val:
        raise ValueError("No file uploaded yet - pick a PDF in the widget first.")
    if isinstance(val, dict):                      # ipywidgets 7.x
        name, info = next(iter(val.items()))
        return name, bytes(info["content"])
    f = val[0]                                     # ipywidgets 8.x
    return f["name"], bytes(f["content"])

filename, pdf_bytes = get_uploaded_pdf(uploader)

reader = PdfReader(io.BytesIO(pdf_bytes))
pages = [p.extract_text() or "" for p in reader.pages]

if not any(t.strip() for t in pages):
    raise ValueError("No text extracted - this may be a scanned PDF that needs OCR.")
```

```python
job_description = """
Overview
As a Software Engineer do you enjoy solving problems, looking at problems through a different lens,
and working closely with customers to innovate new solutions to complex problems? Do you jump with
excitement at the opportunity to identify trends and provide unique business solutions? Do you want
to join a team where learning about a new technology or solution is part of our work every day?

Responsibilities
- Collaborates with appropriate stakeholders to determine user requirements for a scenario.
- Drives identification of dependencies and the development of design documents for a product,
  application, service, or platform.
- Creates, implements, optimizes, debugs, refactors, and reuses code to establish and improve
  performance and maintainability, effectiveness, and return on investment (ROI).
- Leverages subject-matter expertise of product features and partners with appropriate stakeholders
  (e.g., project managers) to drive a workgroup's project plans, release plans, and work items.
- Acts as a Designated Responsible Individual (DRI) and guides other engineers by developing and
  following the playbook, working on call to monitor system/product/service for degradation,
  downtime, or interruptions, alerting stakeholders about status and initiating actions to restore
  system/product/service for simple and complex problems when appropriate.
- Proactively seeks new knowledge and adapts to new trends, technical solutions, and patterns that
  will improve the availability, reliability, efficiency, observability, and performance of products
  while also driving consistency in monitoring and operations at scale.

Qualifications
- Bachelor and/or Graduate degree in computer science, engineering or equivalent.
- Experience: 2+ years of professional development.
- Technical Stack: Python, C++, C#, ML, NLP, and search technologies.
- Focus Areas: Agentic Development, strong analytical skills, customer engagement, and
  cross-functional collaboration.
- Familiarity with Microsoft Cloud Services (Azure, Entra, O365).

Other Requirements
- Ability to meet Microsoft, customer and/or government security screening requirements.

Preferred/Additional Qualifications
- Self-identifies as a technical software engineer with a passion for cloud technologies and AI.
- Previous experience as a Software Developer Engineer.
- Track record of working closely with agile and diverse engineering teams.
- Understanding of cloud, AI and other computing technologies, including business drivers and
  emerging trends across Microsoft and non-Microsoft technologies.
- Experience working in geographically dispersed teams.
- Self-starter who likes tackling new challenges and working independently.
- Ability to build relationships with senior technical and business leaders.
- Strong verbal and written communication ability including fluency in English.
"""

state = {
    "job_description": job_description.strip(),
    "resume": {
        "filename": filename,
        "text": "\n\n".join(pages),
    },
}

classifier = TypeSafeClassifier()

response = classifier.invoke({
    "state": state,
    "questions": {
        "is_good_match": Noul(
            instructions=(
                "The candidate described in the resume is a strong match for the "
                "job_description, meeting most required qualifications."
            )
        )
    },
})

for key, result in response.nouls.items():
    print(f"{key:22s} {result.noul:.2f}")

match_prob = response.nouls["is_good_match"].noul
print("\nGood match" if match_prob >= 0.5 else "\nNot a strong match", f"({match_prob:.0%})")
```

# LLM

```python
from pydantic import BaseModel, Field
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

class MatchResult(BaseModel):
    is_good_match: bool = Field(description="True if the candidate meets most required qualifications.")
    confidence: float = Field(
        ge=0.0, le=1.0,
        description=(
            "How confident you are in the is_good_match verdict, from 0.0 to 1.0. "
            "Use high values (>0.85) only when the resume gives clear, direct evidence for or against "
            "the key requirements. Use lower values (0.5-0.7) when the resume is vague, missing "
            "details, or the candidate is borderline."
        ),
    )
    
llm = init_chat_model("gpt-5.5")
structured_llm = llm.with_structured_output(MatchResult)

prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are an experienced technical recruiter. Compare a candidate's resume against a job "
     "description and judge fit strictly on evidence in the resume. Do not assume skills that "
     "are not stated. Weigh required qualifications more heavily than preferred ones."),
    ("human",
     "<job_description>\n{job_description}\n</job_description>\n\n"
     "<resume filename=\"{filename}\">\n{resume}\n</resume>\n\n"
     "Is this candidate a good match for the job?"),
])

chain = prompt | structured_llm

result: MatchResult = chain.invoke({
    "job_description": job_description.strip(),
    "resume": "\n\n".join(pages),
    "filename": filename,
})

print(f"Good match:  {result.is_good_match}  (confidence: {result.confidence:.0%})")
```
